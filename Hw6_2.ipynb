 РЕАЛІЗАЦІЯ МЕТОДУ ГРАДІЄНТНОГО СПУСКУ ТА АНАЛІТИЧНОГО РІШЕННЯ

```python
def h(X,w):
  return np.dot(X,w)

def loss_function(X,y,w):
  return np.square(h(X,w)-y).sum()/(2*len(y))

def gradient_step(X,y,w,learning_rate):
  m=len(y)
  grad=(X.T@(h(X,w)-y))/m
  w=w-learning_rate*grad
  return w

```

```python
def gradient(X,y,learning_rate,num_iter,eps):
  ones=np.ones((X.shape[0],1))
  X=np.hstack((ones,X))
  w=np.zeros(X.shape[1])
  loss=loss_function(X,y,w)
  loss_history=[loss]
  for _ in range(num_iter):
    w=gradient_step(X,y,w,learning_rate)
    loss=loss_function(X,y,w)
    if abs(loss-loss_history[-1])<eps:
      loss_history.append(loss)
      break
    loss_history.append(loss)
  return w,loss_history


norm=df.copy()
columns=['price','area','bedrooms','bathrooms']
for column in columns:
  norm[column]=(df[column]-df[column].mean())/df[column].std()
X=norm[['area','bedrooms','bathrooms']].values
y=norm['price'].values
learning_rate=0.001
num_iter=10000
eps=0.00000001
w,loss_history=gradient(X,y,learning_rate,num_iter,eps)
ones=np.ones((X.shape[0],1))
X=np.hstack((ones,X))
w_analitic=np.linalg.inv(X.T@X)@X.T@y
print('Gradient_Descent=',w.round(4),';loss_function=',loss_function(X,y,w).round(4))
print('Analitic_Solve=  ',w_analitic.round(4),';loss_function=',loss_function(X,y,w_analitic).round(4))
```
